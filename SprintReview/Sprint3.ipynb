```
public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price) { Name = name; Price = price; }

    public virtual string GetInfo() => $"Компонент: {Name}, Цена: {Price} руб.";
    public override string ToString() => GetInfo();

    public static bool operator ==(Component comp1, Component comp2) => 
        ReferenceEquals(comp1, comp2) || (!(comp1 is null) && !(comp2 is null) && comp1.Name == comp2.Name && comp1.Price == comp2.Price);
    public static bool operator !=(Component comp1, Component comp2) => !(comp1 == comp2);
    public override bool Equals(object obj) => obj is Component other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public interface IInstallable { void Install(); }

public class CPU : Component, IInstallable
{
    public double Frequency { get; set; }
    public int Cores { get; set; }
    public CPU(string name, decimal price, double frequency, int cores) : base(name, price) { Frequency = frequency; Cores = cores; }
    public override string GetInfo() => $"CPU: {Name}, Ядра: {Cores}, Частота: {Frequency} GHz, Цена: {Price} руб.";
    public void Install() => Console.WriteLine($"Установка процессора {Name} в сокет...");
}

public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public string Type { get; set; }
    public RAM(string name, decimal price, int capacity, string type) : base(name, price) { Capacity = capacity; Type = type; }
    public override string GetInfo() => $"RAM: {Name}, Объем: {Capacity} GB, Тип: {Type}, Цена: {Price} руб.";
    public void Install() => Console.WriteLine($"Установка оперативной памяти {Name} в слот DIMM...");
}

public class GPU : Component, IInstallable
{
    public int VRAM { get; set; }
    public string InterfaceType { get; set; }
    public GPU(string name, decimal price, int vram, string interfaceType) : base(name, price) { VRAM = vram; InterfaceType = interfaceType; }
    public override string GetInfo() => $"GPU: {Name}, Видеопамять: {VRAM} GB, Интерфейс: {InterfaceType}, Цена: {Price} руб.";
    public void Install() => Console.WriteLine($"Установка видеокарты {Name} в слот {InterfaceType}...");
}

public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }
    public string FormFactor { get; set; }
    public HDD(string name, decimal price, int capacity, string formFactor) : base(name, price) { Capacity = capacity; FormFactor = formFactor; }
    public override string GetInfo() => $"HDD: {Name}, Объем: {Capacity} GB, Форм-фактор: {FormFactor}, Цена: {Price} руб.";
    public void Install() => Console.WriteLine($"Установка жесткого диска {Name} в отсек для {FormFactor}...");
}

public class Computer
{
    private List<Component> components = new List<Component>();
    public void AddComponent<T>(T component) where T : Component, IInstallable { components.Add(component); component.Install(); Console.WriteLine($"Компонент {component.Name} добавлен."); }
    public void RemoveComponent<T>(T component) where T : Component { if (components.Remove(component)) Console.WriteLine($"Компонент {component.Name} удален."); else Console.WriteLine($"Компонент {component.Name} не найден."); }
    public decimal GetTotalPrice() => components.Sum(c => c.Price);
    public void PrintAllComponents() { Console.WriteLine("Компоненты компьютера:"); components.ForEach(c => Console.WriteLine($"  {c.GetInfo()}")); Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб."); }
    public List<Component> GetComponents() => components;
}

public static class ComputerFactory
{
    public static Computer CreateGamingPC() { var pc = new Computer(); Console.WriteLine("Создание игрового компьютера..."); pc.AddComponent(new CPU("Intel Core i9-14900K", 55000, 5.8, 24)); pc.AddComponent(new RAM("Corsair Dominator", 12000, 32, "DDR5")); pc.AddComponent(new GPU("NVIDIA RTX 4090", 180000, 24, "PCIe 4.0")); pc.AddComponent(new HDD("Samsung 990 Pro", 15000, 2000, "M.2")); return pc; }
    public static Computer CreateOfficePC() { var pc = new Computer(); Console.WriteLine("Создание офисного компьютера..."); pc.AddComponent(new CPU("Intel Core i5-13400", 25000, 4.6, 10)); pc.AddComponent(new RAM("Kingston Fury", 6000, 16, "DDR4")); pc.AddComponent(new GPU("Integrated Graphics", 0, 0, "Onboard")); pc.AddComponent(new HDD("WD Blue SN580", 8000, 1000, "M.2")); return pc; }
}



Console.WriteLine("1. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА:");
new List<Component> { new CPU("AMD Ryzen 7", 45000, 5.0, 8), new RAM("G.Skill", 9000, 32, "DDR5"), new GPU("AMD RX 7900", 120000, 24, "PCIe 4.0"), new HDD("Seagate", 5000, 4000, "3.5\"") }.ForEach(c => Console.WriteLine(c.GetInfo()));

Console.WriteLine("\n2. ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ ОПЕРАТОРОВ:");
var cpu1 = new CPU("Intel Core i7", 30000, 4.8, 8);
var cpu2 = new CPU("Intel Core i7", 30000, 4.8, 8);
Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");

Console.WriteLine("\n3. ДЕМОНСТРАЦИЯ РАБОТЫ С КОМПЬЮТЕРОМ:");
var myComputer = new Computer();
myComputer.AddComponent(new CPU("Intel Core i5", 22000, 4.9, 10));
myComputer.AddComponent(new RAM("Crucial", 7000, 16, "DDR4"));
myComputer.AddComponent(new GPU("NVIDIA RTX 4070", 75000, 12, "PCIe 4.0"));
myComputer.PrintAllComponents();

Console.WriteLine("\n4. ДЕМОНСТРАЦИЯ УДАЛЕНИЯ КОМПОНЕНТА:");
myComputer.RemoveComponent(new GPU("NVIDIA RTX 4070", 75000, 12, "PCIe 4.0"));
myComputer.PrintAllComponents();

Console.WriteLine("\n5. ДЕМОНСТРАЦИЯ ФАБРИКИ:");
var gamingPC = ComputerFactory.CreateGamingPC();
gamingPC.PrintAllComponents();

Console.WriteLine("\n6. ДЕМОНСТРАЦИЯ ИНТЕРФЕЙСА:");
new List<IInstallable> { new CPU("AMD Ryzen 9", 60000, 5.7, 16), new RAM("Corsair", 11000, 64, "DDR5") }.ForEach(c => c.Install());

```
