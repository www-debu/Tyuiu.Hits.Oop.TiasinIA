```

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void Printing()
    {
        Console.WriteLine($"Сотрудник: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary:C}");
    }
}


public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department) 
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void Printing()
    {
        base.Printing();
        Console.WriteLine($"Отдел: {Department}");
        Console.WriteLine("---");
    }
}


public class ProjectManager : Manager
{
    public string ProjectName { get; set; }

    public ProjectManager(string name, string position, double salary, string department, string projectName) 
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    public override void Printing()
    {
        base.Printing();
        Console.WriteLine($"Проект: {ProjectName}");
        Console.WriteLine("---");
    }
}


public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}


public class AdvancedProjectManager : ProjectManager, IReportable, ITeamLeader
{
    private List<string> teamMembers;

    public AdvancedProjectManager(string name, string position, double salary, string department, string projectName) 
        : base(name, position, salary, department, projectName)
    {
        teamMembers = new List<string>();
    }

    
    public string GenerateReport()
    {
        return $"Отчет по проекту '{ProjectName}'. " +
               $"Менеджер: {Name}. " +
               $"Отдел: {Department}. " +
               $"Команда: {teamMembers.Count} человек.";
    }

    
    public string[] GetTeamMembers()
    {
        return teamMembers.ToArray();
    }

    
    public void AddTeamMember(string memberName)
    {
        teamMembers.Add(memberName);
        Console.WriteLine($"Добавлен член команды: {memberName}");
    }

    public void RemoveTeamMember(string memberName)
    {
        if (teamMembers.Remove(memberName))
        {
            Console.WriteLine($"Удален член команды: {memberName}");
        }
    }

    public override void Printing()
    {
        base.Printing();
        Console.WriteLine($"Размер команды: {teamMembers.Count} человек");
        Console.WriteLine("---");
    }
}
```

```

display("=== ТЕСТИРОВАНИЕ СТРУКТУРЫ КОМПАНИИ ===");


var employee = new Employee("Иван Петров", "Разработчик", 80000);
var manager = new Manager("Анна Сидорова", "Менеджер отдела", 120000, "IT");
var projectManager = new ProjectManager("Сергей Иванов", "Проектный менеджер", 150000, "Разработка", "Новый веб-сайт");

var advancedPM = new AdvancedProjectManager(
    "Мария Козлова", 
    "Старший проектный менеджер", 
    180000, 
    "Разработка продуктов", 
    "Мобильное приложение"
);


advancedPM.AddTeamMember("Алексей Смирнов");
advancedPM.AddTeamMember("Елена Волкова");
advancedPM.AddTeamMember("Дмитрий Орлов");

display("1. БАЗОВЫЙ СОТРУДНИК:");
employee.Printing();

display("2. МЕНЕДЖЕР:");
manager.Printing();

display("3. ПРОЕКТНЫЙ МЕНЕДЖЕР:");
projectManager.Printing();

display("4. РАСШИРЕННЫЙ ПРОЕКТНЫЙ МЕНЕДЖЕР:");
advancedPM.Printing();
```

```

display("=== ТЕСТИРОВАНИЕ ИНТЕРФЕЙСОВ ===");


var report = advancedPM.GenerateReport();
display($"Отчет: {report}");


var team = advancedPM.GetTeamMembers();
display("Члены команды:");
foreach (var member in team)
{
    display($"  - {member}");
}


display("=== ДОПОЛНИТЕЛЬНЫЕ ТЕСТЫ ===");


advancedPM.RemoveTeamMember("Елена Волкова");


var updatedReport = advancedPM.GenerateReport();
display($"Обновленный отчет: {updatedReport}");


var updatedTeam = advancedPM.GetTeamMembers();
display("Обновленный состав команды:");
foreach (var member in updatedTeam)
{
    display($"  - {member}");
}
```

```

display("=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");

List<Employee> employees = new List<Employee>
{
    new Employee("Петр Николаев", "Аналитик", 70000),
    new Manager("Ольга Васнецова", "Руководитель отдела", 130000, "Маркетинг"),
    new ProjectManager("Андрей Семенов", "Проектный менеджер", 160000, "Продажи", "Увеличение клиентской базы"),
    advancedPM
};

display($"Всего сотрудников в системе: {employees.Count}");

foreach (var emp in employees)
{
    display($"--- Информация о сотруднике: {emp.Name} ---");
    emp.Printing();
    
    
    if (emp is IReportable reportable)
    {
        var empReport = reportable.GenerateReport();
        display($"📊 Отчет: {empReport}");
    }
    
    if (emp is ITeamLeader teamLeader)
    {
        var members = teamLeader.GetTeamMembers();
        display($"👥 Размер команды: {members.Length} человек");
    }
    
    display(""); // Пустая строка для разделения
}
```
